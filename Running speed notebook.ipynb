```python
# plot running aligned to start of hit trials

f = h5py.File(path, "r")
rotaryEncoderCount=f['rotaryEncoderCount'][:]
rotaryEncoderCountsPerRev=f['rotaryEncoderCountsPerRev'][()]
wheelRadius=f['wheelRadius'][()]
trialResponse=f['trialResponse'][:]
trialStimStartFrame=f['trialStimStartFrame'][:]
trialResponseIndex=np.where(trialResponse==1)[0]
trialNonResponseIndex=np.where(trialResponse==0)[0]

tbins=np.arange(-1,5,1/60)

response_aligned_speed=[]
non_response_aligned_speed=[]
for tt in trialResponseIndex:
    trial_distance=rotaryEncoderCount[trialStimStartFrame[tt]-60:trialStimStartFrame[tt]+300]
    response_aligned_speed.append(np.diff(trial_distance))
for tt in trialNonResponseIndex:
    trial_distance=rotaryEncoderCount[trialStimStartFrame[tt]-60:trialStimStartFrame[tt]+300]
    non_response_aligned_speed.append(np.diff(trial_distance))

response_aligned_speed=np.vstack(response_aligned_speed)
non_response_aligned_speed=np.vstack(non_response_aligned_speed)

fig,ax=plt.subplots(1,1)
ax.axvline(0,color='k',label='stim onset')
ax.plot(tbins[:-1],np.nanmean(response_aligned_speed,axis=0),color='r',label='response trials')
ax.plot(tbins[:-1],np.nanmean(non_response_aligned_speed,axis=0),color='b',label='non-response trials')
ax.set_xlabel('time relative to stimulus onset (s)')
ax.set_ylabel('speed (arbitrary units)')
ax.set_title('running speed aligned to stimulus onset')
ax.legend()
f.close()
```
